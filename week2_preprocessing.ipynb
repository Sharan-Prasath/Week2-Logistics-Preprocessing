```python
import pandas as pd

df = pd.read_csv('dynamic_supply_chain_logistics_dataset_with_country.csv')
```

```python
print(df.shape)
print(df.columns)
df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
print(df.isnull().sum())
```

```python
df.fillna(df.median(numeric_only=True), inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)

df = df.drop_duplicates()
```

```python
Q1 = df['shipping_costs'].quantile(0.25)
Q3 = df['shipping_costs'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['shipping_costs'] < lower) |
              (df['shipping_costs'] > upper)]

print("Outliers:", len(outliers))
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

cols = [
    'warehouse_inventory_level',
    'shipping_costs',
    'supplier_reliability_score',
    'lead_time_days',
    'historical_demand'
]

df[cols] = scaler.fit_transform(df[cols])
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['risk_classification'] = le.fit_transform(
    df['risk_classification']
)

df['supplier_country'] = le.fit_transform(
    df['supplier_country']
)
```

```python
df.to_csv("cleaned_logistics_dataset.csv", index=False)
```
